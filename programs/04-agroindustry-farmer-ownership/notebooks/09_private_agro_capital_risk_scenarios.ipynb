# Agroindustria privada: escenarios de capital, rentabilidad y agua
**Simulación sintética, no estudio de mercado.** Moneda: millones COP constantes. La rotación 15/15 se modela con dos cuadrillas y NO constituye validación de jornada laboral.


```python
import sys
sys.path.insert(0,'../python')
from dataclasses import replace
from colombia_dev.agro_finance import Project,evaluate
import matplotlib.pyplot as plt
```

```python
p=Project()
r=evaluate(p)
print('NPV (millones COP):',round(r['npv_mcop'],1))
print('EBITDA primer año:',round(r['year1']['ebitda_mcop'],1))
print('Uso de agua m3/año:',r['annual_water_m3'],'cumple techo?',r['water_within_cap'])
```

```python
cases={'Base':p,'Precio -20%':replace(p,sale_price_mcop_ton=p.sale_price_mcop_ton*.8),'Rendimiento -25%':replace(p,yield_ton_ha=p.yield_ton_ha*.75),'CAPEX +35%':replace(p,capex_mcop=p.capex_mcop*1.35),'Costo capital +4pp':replace(p,discount_rate=p.discount_rate+.04),'Precio +35%':replace(p,sale_price_mcop_ton=p.sale_price_mcop_ton*1.35)}
results={k:evaluate(v) for k,v in cases.items()}
fig,ax=plt.subplots(figsize=(10,5))
ax.barh(list(results),[x['npv_mcop'] for x in results.values()])
ax.axvline(0,linewidth=1)
ax.set(xlabel='VPN (millones COP)',title='Sensibilidad del capital privado — supuestos sintéticos')
ax.grid(axis='x',alpha=.25);plt.show()
```

## Interpretación
Un VPN negativo obliga a rediseñar el proyecto antes de buscar equity. El retorno exigido por el inversionista debe compararse con la capacidad real de generar flujo libre, después de remunerar trabajo, productores, transporte y mantenimiento. Antes de implementar se requieren salarios legales detallados, calendario de cosecha, deuda y amortización, impuestos reales, reservas, derechos de propiedad y datos territoriales observados.
