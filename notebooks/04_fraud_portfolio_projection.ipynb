# Proyección de riesgo operativo antifraude\nEscenarios sintéticos de solicitudes, prevalencia y pérdida por fraude. Este notebook **no** estima tasas reales ni calibra modelos.\nSe calcula pérdida esperada anual y distribución simulada de conteos con Binomial mediante muestreo Bernoulli.\n

```python
import random\nimport matplotlib.pyplot as plt\n
```

```python
seed=2026; rng=random.Random(seed)\nrequests=1200; avg_loss=750000; scenarios={'controlado':.005,'base':.015,'estres':.04}\nlosses={}\nfor name, prevalence in scenarios.items():\n    losses[name]=[sum(rng.random()<prevalence for _ in range(requests))*avg_loss for _ in range(600)]\nfor name, sample in losses.items():\n    print(name, 'mean_loss_COP', round(sum(sample)/len(sample)), 'expected_formula_COP', round(requests*scenarios[name]*avg_loss))\n
```

```python
fig, ax=plt.subplots(figsize=(9,5))\nax.boxplot([losses[k] for k in scenarios], tick_labels=list(scenarios))\nax.set(title='Distribución simulada de pérdidas por escenario', ylabel='Pérdida anual sintética (COP)')\nax.grid(axis='y',alpha=.25)\nplt.show()\n
```
