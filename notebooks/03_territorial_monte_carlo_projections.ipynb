# Proyecciones territoriales bajo incertidumbre\nSimulación Monte Carlo geométrica de un **índice sintético** de capacidad/actividad territorial (base 100). No representa datos oficiales de Colombia.\n\n**Hipótesis:** crecimiento esperado anual 3.5%, volatilidad 12%; horizontes 10 años, 5,000 trayectorias. Evaluar sensibilidad y bandas p05/p50/p95.\n

```python
import sys\nsys.path.insert(0, '../python')\nfrom colombia_dev.projections import ProjectionConfig, project\nimport matplotlib.pyplot as plt\n
```

```python
cfg = ProjectionConfig()\nrows = project(cfg)\nyears = [r['year'] for r in rows]\nmedian = [r['p50'] for r in rows]\nlow = [r['p05'] for r in rows]\nhigh = [r['p95'] for r in rows]\nfig, ax = plt.subplots(figsize=(10,5))\nax.plot(years, median, label='Mediana (p50)', linewidth=2)\nax.fill_between(years, low, high, alpha=.22, label='Intervalo 5%-95%')\nax.set(xlabel='Años desde el inicio', ylabel='Índice sintético (base 100)', title='Proyección territorial Monte Carlo')\nax.grid(alpha=.25)\nax.legend()\nplt.show()\n
```
