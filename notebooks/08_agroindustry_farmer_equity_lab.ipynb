```python
import sys\nsys.path.insert(0,'../python')\nfrom colombia_dev.agro_equity import Ownership,AgroScenario,operating_result,equity_after_round,hypothetical_distribution\nimport matplotlib.pyplot as plt\n
```

```python
own=Ownership()\nbase=AgroScenario(hectares=100,yield_tons_per_ha=8,sale_price_per_ton=1.2,farm_cost_per_ha=3.5,processing_cost_per_ton=.18,fixed_cost=110,spoilage_rate=.08)\nfor shock in (-.25,0,.25):\n    case=AgroScenario(**{**base.__dict__,'sale_price_per_ton':base.sale_price_per_ton*(1+shock)})\n    print('price shock',shock,'result',operating_result(case))\n
```

```python
rounds=[0,10,25,50,100]\nfarmer_share=[equity_after_round(own,100,new)['farmers'] for new in rounds]\nfig,ax=plt.subplots(figsize=(9,4))\nax.plot(rounds,[100*x for x in farmer_share],marker='o')\nax.set(xlabel='Nueva inversión (unidades monetarias)',ylabel='Participación de productores (%)',title='Dilución bajo una ronda primaria hipotética')\nax.grid(alpha=.25);plt.show()\n
```

```python
out=operating_result(base)\nsurplus=max(0,out['operating_surplus_before_tax_finance'])\nprint('Excedente operativo:',surplus)\nprint('Distribución puramente ilustrativa:',hypothetical_distribution(surplus,.3,own))\n
```
