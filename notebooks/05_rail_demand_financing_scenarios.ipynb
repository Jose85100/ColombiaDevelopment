```python
import sys\nsys.path.insert(0, '../python')\nfrom colombia_dev.rail import RailScenario, evaluate\nimport matplotlib.pyplot as plt\n
```

```python
base=RailScenario(annual_tons=850,passengers=450,freight_rate_per_ton=.55,fare_per_passenger=.15,operating_cost=290,capex=2800,discount_rate=.10,horizon=25,annual_growth=.025)\nscenarios={'pesimista': RailScenario(**{**base.__dict__,'annual_growth':0,'annual_tons':550}), 'base':base, 'optimista':RailScenario(**{**base.__dict__,'annual_growth':.05,'annual_tons':1200})}\nresults={k:evaluate(v) for k,v in scenarios.items()}\nfor k,v in results.items(): print(k,round(v['npv'],2))\n
```

```python
fig,ax=plt.subplots(figsize=(10,5))\nfor name,result in results.items():\n    cumulative=[sum(result['cashflows'][:i+1]) for i in range(len(result['cashflows']))]\n    ax.plot(range(len(cumulative)),cumulative,label=name)\nax.axhline(0,linestyle='--',linewidth=1)\nax.set(xlabel='Años',ylabel='Flujo acumulado sin descuento (unidades monetarias)',title='Escenarios de recuperación simple')\nax.grid(alpha=.25);ax.legend();plt.show()\n
```
