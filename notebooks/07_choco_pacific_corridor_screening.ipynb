```python
options=[Corridor('Montaña-túneles',.85,.35,.70,.35,.55,.80),Corridor('Atrato-intermodal',.50,.65,.35,.75,.65,.70),Corridor('Rodeo-superficie',.65,.80,.40,.35,.40,.55),Corridor('Zona excluida',.20,.20,.15,.20,.95,.90,True)]\nranked=rank_corridors(options)\nranked\n
```

```python
valid=[r for r in ranked if r['feasible']]\nfig,ax=plt.subplots(figsize=(9,4))\nax.barh([r['name'] for r in valid],[r['penalty'] for r in valid])\nax.invert_yaxis();ax.set(xlabel='Penalización multicriterio (menor es mejor)',title='Cribado ilustrativo de alternativas')\nax.grid(axis='x',alpha=.25);plt.show()\n
```
